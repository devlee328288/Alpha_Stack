## 해석 범위

- Balanced Accuracy와 MCC는 불균형한 세 클래스의 예측 균형을 확인한다.
- PR-AUC는 저장된 결과에 OOS 클래스 확률 배열이 없어 이 요약본에서는 미측정이다.
- MDD와 Sharpe Ratio는 포지션·거래비용 규칙이 정해진 뒤 계산한다.
- 동일 개발 OOS에서 선정했으므로 최종 채택 전 별도 홀드아웃 검증이 필요하다.
