선정 실험: **기본 조합**  
선정 기준: 전체 OOS Accuracy, Macro F1, 하락 Recall의 조화평균  
원본 결과: [03.XGBoost.ipynb](../../조합D_option2_volatility_focus/03.XGBoost.ipynb)


## 선정 이유

같은 조합과 모델의 기본·Daily Return·5Day Return·두 수익률 추가 실험을 비교하고 OOS 핵심지표 조화평균이 가장 높은 결과를 선정했다. 한 지표가 0이면 조화평균도 0이므로 하락을 전혀 예측하지 못한 실험은 best로 선택되기 어렵다.

| 실험 | Accuracy | Macro F1 | 하락 Recall | 핵심지표 조화평균 |
|---|---:|---:|---:|---:|
| **기본 조합 (선정)** | 0.3611 | 0.3542 | 0.3039 | **0.3377** |
| 조합D + 5Day_Return | 0.3514 | 0.3450 | 0.2892 | **0.3260** |
| 조합D + Daily_Return | 0.3472 | 0.3422 | 0.2941 | **0.3260** |
| 조합D + Daily_Return + 5Day_Return | 0.3389 | 0.3319 | 0.2745 | **0.3123** |

## 해석 범위

- Balanced Accuracy와 MCC는 불균형한 세 클래스의 예측 균형을 확인한다.
- PR-AUC는 저장된 결과에 OOS 클래스 확률 배열이 없어 이 요약본에서는 미측정이다.
- MDD와 Sharpe Ratio는 포지션·거래비용 규칙이 정해진 뒤 계산한다.
- 동일 개발 OOS에서 선정했으므로 최종 채택 전 별도 홀드아웃 검증이 필요하다.
